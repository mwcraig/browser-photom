# Local images via the local helper

This notebook loads FITS images from your disk through the **local helper**,
which serves a directory over `http://localhost:8001` and CORS-proxies
astronomy services. Works in any browser; the notebook stays in the default
JupyterLite drive.

Before running, start the helper in a terminal on the host:

```sh
pixi run helper ~/Downloads/ey_uma
```

````python
import helper

helper.list_images()
````

````python
hdul = helper.open_fits("ey-uma-S001-R001-C001-rp.fit")
hdul.info()
````

````python
from astrowidgets.bqplot import ImageWidget

imw = ImageWidget()
imw.load_image(hdul[0].data)
imw
````

````python
imw.cuts = "95%"
````
